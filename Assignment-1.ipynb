```python
#Assignment 1
[1] Assign 5 elements in the dictionary named my_dict. This dictionary contains students information. Add and delete student information from my_dict. Display student information.

my_dict = {
    "Kathan": [1, "CSE AIDS", 75],
    "Raghav":[2,"CSE AIDS", 90],
    "Preet":[3,"CORE CSE",23],
    "Yuvraj":[4,"BioChemical",40],
    "Rudra":[5,"Law",39]
}

my_dict["Ronaldo"] = [1, 75, 25] # Adding student information
del my_dict["Kathan"] # Removing student information

print(my_dict)
{'Raghav': [2, 'CSE AIDS', 90], 'Preet': [3, 'CORE CSE', 23], 'Yuvraj': [4, 'BioChemical', 40], 'Rudra': [5, 'Law', 39], 'Shavez': [1, 75, 25]}
[2] Convert list to dictionary

    list1 = ['name', 'panel', 'rollno']
    list2 = ['ABC', 'B', 34]
    my_dict = {'name': 'ABC', 'panel': 'B', 'rollno': 34}
list1 = ['name', 'panel', 'rollno']
list2 = ['ABC', 'B', 34]
my_dict = {}

for i in range(3):
    my_dict[list1[i]] = list2[i]

print(my_dict)
{'name': 'ABC', 'panel': 'B', 'rollno': 34}
[3] Sort the elements of the dictionary using sorted() function.

my_dict = {'name': 'ABC', 'panel': 'B', 'rollno': 34, 'marks': [65, 87, 67, 94]}
sorted(my_dict)
['marks', 'name', 'panel', 'rollno']
[4] Convert a dictionary to lists. Make one list of keys and other of values.

my_dict = {'name': 'ABC', 'panel': 'B', 'rollno': 34, 'marks': [65, 87, 67, 94]}
keys = list(my_dict.keys())
values = list(my_dict.values())

print("Keys = ", keys)
print("Value = ", values)
Keys =  ['name', 'panel', 'rollno', 'marks']
Value =  ['ABC', 'B', 34, [65, 87, 67, 94]]
[5] Find the mean of values of dictionary.

my_dict = {
    'marks1': 23,
    'marks2': 125,
    'marks3': 43,
    'marks4': 13,
    'marks5': 39
}

average = 0

for key in my_dict:
    average += my_dict[key]

average /= len(my_dict)

print(average)
48.6
[6] my_dict = {'name': ['Aryan', 'Manthan', 'Messi'], 'rollno': [21, 22, 23], 'marks': [78, 69, 80]}
Perform the following operations on above dictionary:
a Display name as Messi
b Display 23 roll no
c Display greatest marks

my_dict = {'name': ['Aryan', 'Manthan', 'Messi'], 'rollno': [21, 22, 23], 'marks': [78, 69, 80]}

# A: Display name as Messi
print(my_dict['name'][3])

# B: Display 22 roll no
print(my_dict['rollno'][1])

# C: Display greatest marks
greatest_marks = my_dict['marks'][0]

for i in range(1, len(my_dict['marks'])):
    if my_dict['marks'][i] > greatest_marks:
        greatest_marks = my_dict['marks'][i]

print(greatest_marks)
Messi
22
80

[7] Calculate frequency of letters occuring in a given string.

string = "Kathan"
frequencies = {}

for c in string:
    if c in frequencies:
        frequencies[c] += 1
    else:
        frequencies[c] = 1

print(frequencies)
{'a': 2, 'd': 1, 'n': 2}


[8] Write a program which takes numerator and denominator values from the user and gives output as a tuple of quotient and remainder.

n = int(input("Enter numerator: "))
d = int(input("Enter denominator: "))

print("Quotient = ", n // d)
print("Remainder = ", n % d)
Quotient =  1
Remainder =  2


[9] Suppose a date is represented as a tupe (d, m, y). Write a program to create two date tuples with values taken from the user and find the number of days between the two dates as an output.

days_in_month = [31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]

d1 = int(input("Enter day of the first date: "))
m1 = int(input("Enter month of the first date: "))
y1 = int(input("Enter year of the first date: "))

d2 = int(input("\nEnter day of the second date: "))
m2 = int(input("Enter month of the second date: "))
y2 = int(input("Enter year of the second date: "))

if (y1, m1, d1) > (y2, m2, d2):
    from_date = (d2, m2, y2)
    to_date = (d1, m1, y1)
else:
    from_date = (d1, m1, y1)
    to_date = (d2, m2, y2)

d1, m1, y1 = from_date
d2, m2, y2 = to_date

days = 0

for y in range(y1, y2):
    if (y % 4 == 0 and y % 100 != 0) or (y % 400 == 0):
        days += 366
    else:
        days += 365

for m in range(1, m1):
    days -= days_in_month[m - 1]
    if m == 2 and ((y1 % 4 == 0 and y1 % 100 != 0) or (y1 % 400 == 0)):
        days -= 1

days -= d1

for m in range(1, m2):
    days += days_in_month[m - 1]
    if m == 2 and ((y2 % 4 == 0 and y2 % 100 != 0) or (y2 % 400 == 0)):
        days += 1

days += d2

print("\nNumber of days between the dates:", days)
Number of days between the dates: 2245
```
