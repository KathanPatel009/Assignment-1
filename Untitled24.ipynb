```python
#create two matrices using nested lists
matrix1=[
    [1,2,3],
    [4,5,6]
]
matrix2=[
    [7,8,9],
    [10,11,12]
]

#initialize a result matrix with the same size 
result=[
    [0,0,0],
    [0,0,0]
]

#Perform matrix addition
for i in range(len(matrix1)):
    for j in range(len(matrix1[0])):
        result[i][j]=matrix1[i][j] + matrix2[i][j]

#Display the result
print("matrix1:")
for row in matrix1:
    print(row)
          
print("\nmatrix2:")
for row in matrix2:
    print(row)

print("\nsum of matrix1 and matrix2:")
for row in result:
    print(row)
```
